```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("BlinkIT Grocery Data.csv")
print("first 5 rows of datset")
print(df.head())
print(df.describe())
print(df.isnull().sum())
print(df.info())
print(df.shape)
print(df.columns)
print(df.duplicated().sum())
```

```python
df.columns = df.columns.str.strip()
print(df.columns)
df["ItemContentKey"].unique()
df["Rating"].unique()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Sales"], bins=20)
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Rating"], bins=10)
plt.title("Distribution of Ratings")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10,5))
df["ItemContentKey"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Item Categories")
plt.xlabel("Item Content")
plt.ylabel("Count")
plt.show()
```

```python
print("Total Sales:", df["Sales"].sum())
print("average Sales:", df["Sales"].mean())
print("maximum Sales:", df["Sales"].max())
print("minimum Sales:", df["Sales"].min())
print("Average Rating:", df["Rating"].mean())
top_items = df.groupby("Item Identifier")["Sales"].sum().sort_values(ascending=False).head(10)

print(top_items)

top_rated = df.groupby("Item Identifier")["Rating"].mean().sort_values(ascending=False).head(10)

print(top_rated)

correlation = df[["Item Visibility","Item Weight","Sales","Rating"]].corr()

print(correlation)

plt.Figure(figsize=(8,6))
sns.heatmap(correlation,annot=True,cmap="coolwarm")
plt.title("heat map coorelation")
plt.show()

```

```python
top_sales = df.groupby("Item Identifier")["Sales"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))
top_sales.plot(kind="bar", color="skyblue")
plt.title("Top 10 Best Selling Items")
plt.xlabel("Item Identifier")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
top_rating = df.groupby("Item Identifier")["Rating"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))
top_rating.plot(kind="bar", color="orange")
plt.title("Top 10 Highest Rated Items")
plt.xlabel("Item Identifier")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Item Visibility"], bins=20)
plt.title("Item Visibility Distribution")
plt.xlabel("Item Visibility")
plt.ylabel("Count")
plt.show()
```

```python
plt.hist(df["Item Weight"],bins=20)
plt.title("item weight distribution")
plt.xlabel("item weight")
plt.ylabel("count")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Rating"], df["Sales"])
plt.title("Sales vs Rating")
plt.xlabel("Rating")
plt.ylabel("Sales")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Item Visibility"], df["Sales"])
plt.title("Sales vs item visibility")
plt.xlabel("Item visibility")
plt.ylabel("sales")
plt.show()
```
